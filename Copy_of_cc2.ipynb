```python
# ============================================================
# CLOUD PARKING MANAGEMENT SYSTEM
# Google Colab - Python
# ============================================================

import pandas as pd
from datetime import datetime

# -----------------------------
# Parking Slot Configuration
# -----------------------------
TOTAL_SLOTS = 10

parking_slots = {
    i: None for i in range(1, TOTAL_SLOTS + 1)
}

# Data stored in the system
parking_records = []

# -----------------------------
# Display Parking Slots
# -----------------------------
def display_slots():
    print("\n========== PARKING SLOTS ==========")

    for slot, vehicle in parking_slots.items():
        if vehicle is None:
            print(f"Slot {slot}: AVAILABLE")
        else:
            print(f"Slot {slot}: OCCUPIED - {vehicle}")

    print("===================================\n")


# -----------------------------
# Park Vehicle
# -----------------------------
def park_vehicle():
    print("\n========== VEHICLE ENTRY ==========")

    vehicle_number = input("Enter vehicle number: ").upper()

    # Check whether vehicle is already parked
    for record in parking_records:
        if record["Vehicle Number"] == vehicle_number and record["Status"] == "Parked":
            print("Vehicle is already parked.")
            return

    # Find an available slot
    available_slot = None

    for slot, vehicle in parking_slots.items():
        if vehicle is None:
            available_slot = slot
            break

    if available_slot is None:
        print("Sorry! Parking is full.")
        return

    # Store vehicle information
    entry_time = datetime.now()

    parking_slots[available_slot] = vehicle_number

    parking_records.append({
        "Vehicle Number": vehicle_number,
        "Slot": available_slot,
        "Entry Time": entry_time,
        "Exit Time": None,
        "Fee": 0,
        "Status": "Parked"
    })

    print("\nVehicle parked successfully!")
    print("Vehicle Number:", vehicle_number)
    print("Parking Slot:", available_slot)
    print("Entry Time:", entry_time.strftime("%Y-%m-%d %H:%M:%S"))


# -----------------------------
# Vehicle Exit
# -----------------------------
def exit_vehicle():
    print("\n========== VEHICLE EXIT ==========")

    vehicle_number = input("Enter vehicle number: ").upper()

    record_found = None

    for record in parking_records:
        if record["Vehicle Number"] == vehicle_number and record["Status"] == "Parked":
            record_found = record
            break

    if record_found is None:
        print("Vehicle not found.")
        return

    exit_time = datetime.now()

    # Calculate parking duration
    duration = exit_time - record_found["Entry Time"]
    hours = duration.total_seconds() / 3600

    # Minimum charge = 1 hour
    charged_hours = max(1, int(hours) + (1 if hours > int(hours) else 0))

    # Parking fee: ₹30 per hour
    fee = charged_hours * 30

    record_found["Exit Time"] = exit_time
    record_found["Fee"] = fee
    record_found["Status"] = "Exited"

    # Free the parking slot
    parking_slots[record_found["Slot"]] = None

    print("\nVehicle exited successfully!")
    print("Vehicle Number:", vehicle_number)
    print("Parking Slot:", record_found["Slot"])
    print("Parking Duration:", round(hours, 2), "hours")
    print("Parking Fee: ₹", fee)


# -----------------------------
# View Parking Records
# -----------------------------
def view_records():
    print("\n========== PARKING RECORDS ==========")

    if len(parking_records) == 0:
        print("No parking records found.")
        return

    df = pd.DataFrame(parking_records)

    print(df.to_string(index=False))


# -----------------------------
# Search Vehicle
# -----------------------------
def search_vehicle():
    print("\n========== SEARCH VEHICLE ==========")

    vehicle_number = input("Enter vehicle number: ").upper()

    found = False

    for record in parking_records:
        if record["Vehicle Number"] == vehicle_number:
            print("\nVehicle Details")
            print("----------------------------")
            print("Vehicle Number:", record["Vehicle Number"])
            print("Slot:", record["Slot"])
            print("Entry Time:", record["Entry Time"])

            if record["Exit Time"] is not None:
                print("Exit Time:", record["Exit Time"])

            print("Fee: ₹", record["Fee"])
            print("Status:", record["Status"])

            found = True

    if not found:
        print("Vehicle not found.")


# -----------------------------
# Parking Statistics
# -----------------------------
def parking_statistics():
    occupied = sum(
        1 for vehicle in parking_slots.values()
        if vehicle is not None
    )

    available = TOTAL_SLOTS - occupied

    total_revenue = sum(
        record["Fee"] for record in parking_records
    )

    print("\n========== PARKING STATISTICS ==========")
    print("Total Parking Slots :", TOTAL_SLOTS)
    print("Occupied Slots      :", occupied)
    print("Available Slots     :", available)
    print("Total Vehicles      :", len(parking_records))
    print("Total Revenue       : ₹", total_revenue)
    print("=========================================")


# -----------------------------
# Main Menu
# -----------------------------
def main():
    while True:

        print("\n")
        print("========================================")
        print("     CLOUD PARKING MANAGEMENT SYSTEM")
        print("========================================")
        print("1. View Parking Slots")
        print("2. Park Vehicle")
        print("3. Vehicle Exit")
        print("4. View Parking Records")
        print("5. Search Vehicle")
        print("6. Parking Statistics")
        print("7. Exit")
        print("========================================")

        choice = input("Enter your choice: ")

        if choice == "1":
            display_slots()

        elif choice == "2":
            park_vehicle()

        elif choice == "3":
            exit_vehicle()

        elif choice == "4":
            view_records()

        elif choice == "5":
            search_vehicle()

        elif choice == "6":
            parking_statistics()

        elif choice == "7":
            print("\nThank you for using Cloud Parking Management System!")
            break

        else:
            print("Invalid choice. Please try again.")


# Start the system
main()
```
